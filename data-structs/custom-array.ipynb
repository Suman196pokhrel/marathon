```python
import ctypes
import sys
```

```python
class MyCustomList:

    def __init__(self):
        self.max_items_allowed = 1
        self.currnet_len = 0
        # Create a C type array with size = self.max_items_allowed
        self.arr = self.__make_array(self.max_items_allowed)

    def __make_array(self,size):
        # Creates a Ctype array (static referential) with size capacity
        return (size * ctypes.py_object)()

    def __len__(self):
        return self.currnet_len

    def __str__(self):
        result = []
        for i in range(self.currnet_len):
            result.append(str(self.arr[i]))

        contents = ", ".join(result)
        return "[" + contents + "]"

    def __iter__(self):
        for i in range(self.currnet_len):
            yield self.arr[i]

    def __getitem__(self, key):
        if not isinstance(key,int):
            return "Invalid key type, only integer types accepted"

        if self.currnet_len == 0:
            raise IndexError("Cannot index an empty list")
    
        # Handling positive keys 
        if key >= 0 :
            if key >= self.currnet_len:
                key = self.currnet_len -1
        # Handling negative keys
        if key < 0:
            if key < -(self.currnet_len):
                key = 0
            else:
                key = self.currnet_len + key

        return self.arr[key]

    def __delitem__(self, key):
        for i in range(key+1, self.currnet_len):
            self.arr[i-1] = self.arr[i]
        self.arr[self.currnet_len - 1] = None
        self.currnet_len -= 1

    def remove(self, item):
        pos = self.find(item)

        if type(pos) == int:
            # delete
            self.__delitem__(pos)

        return -1
                
        

    def find(self,item):
        for i in range(self.currnet_len):
            if self.arr[i] == item:
                return i

        return "error"

    def __resize(self, new_size):
        # Allocate a new memory block with double the memory
        new_Arr = self.__make_array(new_size)
        self.max_items_allowed = new_size

        # Copy elements of existing array to the new one
        for i in range(self.currnet_len):
            new_Arr[i] = self.arr[i]

        self.arr = new_Arr
        return


    def append(self,item):
        if self.currnet_len == self.max_items_allowed:
            self.__resize(self.max_items_allowed * 2)

        self.arr[self.currnet_len] = item
        self.currnet_len += 1
        return

    def pop(self):
        if self.currnet_len == 0 :
            return "Empty list"

        last_index = self.currnet_len - 1
        res = self.arr[last_index]

        self.arr[last_index] = None
        self.currnet_len -= 1
        return res

    

    def clear(self):
        for i in range(self.currnet_len):
            self.arr[i] = None

        self.currnet_len = 0
        self.max_items_allowed = 1


    def insert_at(self, idx, item):
        # check if the size of existing array allows insertion
        if self.currnet_len == self.max_items_allowed:
            self.__resize(self.max_items_allowed * 2)

        # If yes , start shifting elements from left to right , stop at one step right from target indx 
        for i in range(self.currnet_len, idx, -1):
            self.arr[i] = self.arr[i - 1]

        self.arr[idx] = item
        self.currnet_len += 1





myobj1 = MyCustomList()
sys.getsizeof(myobj1)
```

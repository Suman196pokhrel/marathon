```python
class Node:

    def __init__(self, value):
        self.data = value
        self.next = None

    def __str__(self):
        nxt = self.next.data if self.next else None
        return "[" + str(self.data) + " | " + f"{nxt}]"


class LinkedList:

    def __init__(self):
        self.head = None
        self.n = 0

    def __len__(self):
        return self.n

    def __str__(self):
        if self.n == 0:
            return ""
        
        top, mid, bot = [], [], []
        curr = self.head
        while curr is not None:
            val = str(curr.data)
            ptr = "*" if curr.next is not None else "None"
            w1, w2 = len(val) + 2, len(ptr) + 2
            top.append("┌" + "─" * w1 + "┬" + "─" * w2 + "┐")
            mid.append(f"│ {val} │ {ptr} │")
            bot.append("└" + "─" * w1 + "┴" + "─" * w2 + "┘")
            curr = curr.next
            

        return "\n".join(["  ".join(top), "->".join(mid) , "  ".join(bot)])

    def insert_head(self, value):
        new_node = Node(value)
        new_node.next = self.head
        self.head = new_node
        self.n += 1
        return

    def insert_at_index(self, idx:int, value):
        if not isinstance(idx, int):
            raise TypeError("index must be an integer")
        if  idx ==0 or idx >= self.n:
            raise IndexError("Invalid index provided")

        current = self.head
        for i in range(idx-1):
            current = current.next

        node = Node(value)
        node.next = current.next
        current.next = node
        self.n += 1

        return 
        
    def append(self, value):
        node = Node(value)

        if len(self) == 0:
            self.insert_head(value)
            return

        current = self.head
        while current.next is not None:
            current = current.next

        current.next = node
        self.n += 1

        return

    def clear(self):
        self.head = None
        self.n = 0
        return

    def delete_head(self):
        if self.n == 0:
            raise IndexError("Deleting from empty list")

        removed = self.head
        self.head = self.head.next
        self.n -= 1
        return removed

    def delete_tail(self):
        if self.n == 0 :
            raise IndexError("Deleting from empty list")

        elif self.n == 1:
            return self.delete_head()

        current = self.head
        while current.next.next is not None:
            current = current.next

        current.next = None
        self.n -= 1
        return

    def delete_by_value(self,value):
        if self.n == 0 :
            raise IndexError("Deleting in an empty linkedlist")

        # Match at head
        if self.head.data == value:
            removed = self.head
            self.head = self.head.next
            removed.next = None
            self.n -= 1
            return

        # Match after head
        current = self.head.next
        prev = self.head
        while current is not None:
            if current.data == value:
                prev.next = current.next
                current.next = None
                self.n -= 1
                return

            prev = current
            current = current.next
            
        return

    def delete_by_index(self,idx):
        if self.n == 0:
            raise IndexError("Indexing on empty linkedlist")

        if idx >= self.n or idx < 0:
            raise IndexError("Index out of range") 

        if idx == 0:
            removed = self.head
            self.head = self.head.next
            self.n -= 1
            return removed

        curr= self.head
        for _ in range(idx-1):
            curr = curr.next

        removed = curr.next
        curr.next = curr.next.next
        self.n -= 1
        return removed

    def search_by_value(self,value):

        curr = self.head
        counter = 0
        while curr is not None:
            if curr.data == value:
                return counter
            counter += 1
            curr = curr.next

        raise ValueError(f"{value}!r not in list")

    def __getitem__(self, key):
        if self.n == 0:
            raise IndexError("Indexing on empty list")

        if key < 0 or key >= self.n :
            raise IndexError("Index out of range")

        curr = self.head
        for _ in range(key):
            curr = curr.next

        return curr.data



sll = LinkedList()

print("--- insert_head ---")
sll.insert_head(11)
sll.insert_head(0)
print(sll, "\nlen:", len(sll))

print("\n--- append ---")
sll.append(22)
sll.append(33)
sll.append(44)
print(sll, "\nlen:", len(sll))

print("\n--- insert_at_index(1, 99) ---")
sll.insert_at_index(1, 99)
print(sll, "\nlen:", len(sll))

print("\n--- search_by_value ---")
print("33 is at index:", sll.search_by_value(33))
print("0 is at index:", sll.search_by_value(0))

print("\n--- __getitem__ ---")
print("index 0:", sll[0])
print("index 2:", sll[2])
print("last index:", sll[len(sll) - 1])

print("\n--- delete_head ---")
print("removed:", sll.delete_head())
print(sll, "\nlen:", len(sll))

print("\n--- delete_tail ---")
print("removed:", sll.delete_tail())
print(sll, "\nlen:", len(sll))

print("\n--- delete_by_value(22) ---")
sll.delete_by_value(22)
print(sll, "\nlen:", len(sll))

print("\n--- delete_by_index(1) ---")
print("removed:", sll.delete_by_index(1))
print(sll, "\nlen:", len(sll))

print("\n--- clear ---")
sll.clear()
print("len after clear:", len(sll))

print("\n--- error cases ---")
try:
    sll.delete_head()
except IndexError as e:
    print("delete_head on empty:", e)

try:
    sll.delete_tail()
except IndexError as e:
    print("delete_tail on empty:", e)

try:
    sll[0]
except IndexError as e:
    print("indexing on empty:", e)

try:
    sll.search_by_value(5)
except ValueError as e:
    print("search_by_value missing:", e)

try:
    sll.insert_at_index("1", 5)
except TypeError as e:
    print("insert_at_index with str:", e)

try:
    sll.insert_at_index(0, 5)
except IndexError as e:
    print("insert_at_index(0) on empty:", e)
```

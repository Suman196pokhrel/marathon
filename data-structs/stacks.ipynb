```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None



class Stack:

    def __init__(self):
        self.top = None
        self.size = 0

    def isEmpty(self):
        return self.top == None

    def push(self, value):
        node = Node(value)
        node.next = self.top
        self.top = node
        self.size += 1
        return


    def pop(self):
        if self.isEmpty():
            raise IndexError("Empty stack")
        data = self.top.data
        self.top = self.top.next
        self.size -= 1
        return data


    def __str__(self):
        res = ""
        curr = self.top
        while curr is not None:
            res += str(curr.data) + " "
            curr = curr.next

        return res

    def peek(self):
        if self.isEmpty():
            raise IndexError("Empty stack")

        return self.top.data


    def string_reversal(self,data):
        for i in data:
            self.push(i)

        res = ""
        curr = self.top
        while curr is not None:
            res +=  self.pop()
            curr = curr.next
        return res



def text_editor_example(inpString, pattern):
    # input and pattern checks here

    uStack = Stack()
    rStack = Stack()

    # Populate stack
    for i in inpString:
        uStack.push(i)

    for i in pattern:
        if i == 'u':
            temp = uStack.pop()
            rStack.push(temp)
        elif i == "r":
            temp = rStack.pop()
            uStack.push(temp)

    res = ""
    while not uStack.isEmpty():
        res = uStack.pop() + res

    print(res)
    



st1 = Stack()
# st1.push(11)
# st1.push(22)
# st1.push(33)
# print(st1)
# print(st1.isEmpty())
# print(st1.peek())
st1.string_reversal("Suman")

text_editor_example("Hello","uuur")
```

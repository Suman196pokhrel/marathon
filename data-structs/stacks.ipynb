```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None



class Stack:

    def __init__(self):
        self.top = None
        self.size = 0

    def isEmpty(self):
        return self.top == None

    def push(self, value):
        node = Node(value)
        node.next = self.top
        self.top = node
        self.size += 1
        return


    def pop(self):
        if self.isEmpty():
            raise IndexError("Empty stack")

        self.top = self.top.next
        self.size -= 1
        return


    def __str__(self):
        res = ""
        curr = self.top
        while curr is not None:
            res += str(curr.data) + " "
            curr = curr.next

        return res

    def peek(self):
        if self.isEmpty():
            raise IndexError("Empty stack")

        return self.top.data



st1 = Stack()
st1.push(11)
st1.push(22)
st1.push(33)
print(st1)
print(st1.isEmpty())
print(st1.peek())
```

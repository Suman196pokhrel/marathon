```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class Stack:

    def __init__(self):
        self.top = None
        self.size = 0

    def isEmpty(self):
        return self.top == None

    def push(self, value):
        node = Node(value)
        node.next = self.top
        self.top = node
        self.size += 1
        return


    def pop(self):
        if self.isEmpty():
            raise IndexError("Empty stack")
        data = self.top.data
        self.top = self.top.next
        self.size -= 1
        return data


    def __str__(self):
        res = ""
        curr = self.top
        while curr is not None:
            res += str(curr.data) + " "
            curr = curr.next

        return res

    def peek(self):
        if self.isEmpty():
            raise IndexError("Empty stack")

        return self.top.data



class SimulatedQueueUsingStacks:
    def __init__(self):
        self.s1 = Stack()
        self.s2 = Stack()

    def enqueue(self, item):
        self.s1.push(item)



    def dequeue(self):

        if self.s2.size == 0:
            # move everything from s1 to s2
            while self.s1.size != 0:
                self.s2.push(self.s1.pop())


        # pop out the top of s2
        item = self.s2.pop()
        return item



simQueue = SimulatedQueueUsingStacks()
simQueue.enqueue(11)
simQueue.enqueue(22)
simQueue.enqueue(33)
simQueue.enqueue(44)
print(simQueue.dequeue())  # Expected: 11
print(simQueue.dequeue())  # Expected: 22

# # Add more while older items are still in the queue
simQueue.enqueue(55)
simQueue.enqueue(66)

print(simQueue.dequeue())  # Expected: 33
print(simQueue.dequeue())  # Expected: 44
print(simQueue.dequeue())  # Expected: 55
print(simQueue.dequeue())  # Expected: 66

# Check removal from an empty queue
try:
    simQueue.dequeue()
except IndexError as error:
    print("Empty queue check:", error)

# # Check that the queue can be reused after becoming empty
simQueue.enqueue(77)
print(simQueue.dequeue())  # Expected: 77
```
